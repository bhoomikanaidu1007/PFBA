```python
# ─────────────────────────────────────────────────────────────────
# ALWAYS put these imports at the TOP of any notebook you create
# ─────────────────────────────────────────────────────────────────

import seaborn as sns              # The Seaborn library (short name: sns)
import matplotlib.pyplot as plt    # Matplotlib — needed to display charts
import pandas as pd                # Pandas — for organising data in tables

print("All libraries imported successfully!")
```

```python
# Set the visual theme — run this once per notebook
sns.set_theme(style="whitegrid")

# ✏️ Try changing "whitegrid" to one of these and re-run:
#    "darkgrid"   → dark background with grid
#    "white"      → clean white, no grid
#    "dark"       → dark background, no grid
#    "ticks"      → white with small tick marks

print("Theme is set! All charts below will use this style.")
```

```python
# ─── YOUR FIRST SEABORN CHART ────────────────────────────────────

# Step 1: The data
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales  = [50000, 70000, 65000, 90000, 85000, 110000]

# Step 2: Draw the chart
sns.lineplot(x=months, y=sales)

# Step 3: Show it
plt.show()
```

```python
# ─── PROFESSIONAL LINE CHART — with title and labels ─────────────

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales  = [50000, 70000, 65000, 90000, 85000, 110000]

# Draw the line
sns.lineplot(x=months, y=sales)

# Add a title and axis labels
# (We use plt. for titles and labels — Seaborn handles the chart itself)
plt.title('Monthly Sales — First Half of Year')   # Chart headline
plt.xlabel('Month')                               # Label for horizontal axis
plt.ylabel('Revenue (₹)')                         # Label for vertical axis

plt.show()
```

```python
# ─── TWO LINES ON ONE CHART ──────────────────────────────────────

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales  = [50000, 70000, 65000, 90000, 85000, 110000]
profit = [10000, 15000, 12000, 20000, 18000,  25000]

# First line — Sales
# label='Sales' gives this line a name for the legend
sns.lineplot(x=months, y=sales, label='Sales')

# Second line — Profit
# linestyle='--' makes this line dashed so the two lines look different
sns.lineplot(x=months, y=profit, label='Profit', linestyle='--')

# Title, labels, and legend
plt.title('Sales vs Profit — January to June')
plt.xlabel('Month')
plt.ylabel('Amount (₹)')
plt.legend()    # This draws the legend box using the label= values above

plt.show()

# ─── WHAT TO NOTICE ──────────────────────────────────────────────
# Seaborn automatically chose two different colours for the two lines
# The legend box shows which colour belongs to which series
```

```python
# ─── SEABORN BAR CHART — Product Sales ───────────────────────────

# Our data
products = ['Laptop', 'Mobile', 'Watch', 'Tablet']
sales    = [85000,    72000,    30000,   60000]

# Draw the bar chart
# palette='Blues_d' colours the bars in shades of blue
sns.barplot(x=products, y=sales, palette='Blues_d')

# ✏️ Try changing 'Blues_d' to 'Greens_d', 'Set2', or 'viridis'

# Add labels
plt.title('Product-wise Sales — Q1')
plt.xlabel('Product')
plt.ylabel('Sales (₹)')

plt.show()
```

```python
# ─── HORIZONTAL BAR CHART — Department Headcount ─────────────────

# Long department names — these would overlap on a vertical bar chart
departments = ['Human Resources', 'Finance & Accounts', 'Operations',
               'Sales & Marketing', 'IT & Technology']
headcount   = [45, 30, 80, 60, 35]

# orient='h' makes the bars go horizontally (left to right)
# Notice: x and y are swapped compared to a vertical bar chart
sns.barplot(x=headcount, y=departments, palette='Set2', orient='h')

plt.title('Employees per Department')
plt.xlabel('Number of Employees')
plt.ylabel('')    # No label needed — the department names are self-explanatory

plt.tight_layout()    # Prevents names from being cut off at the edges
plt.show()
```

```python
# ─── SEABORN BOX PLOT — Salary Distribution ──────────────────────

# We need numpy to generate sample data (random numbers)
import numpy as np
np.random.seed(42)    # This makes sure everyone gets the same random numbers

# Create a table with 120 rows — 30 employees per department
# np.random.normal(average, spread, count) generates realistic salary data
salary_data = pd.DataFrame({
    'Department': (['Sales'] * 30 + ['HR'] * 30 + ['Finance'] * 30 + ['IT'] * 30),
    'Salary_K':   (
        list(np.random.normal(55, 8,  30)) +   # Sales: average ₹55,000
        list(np.random.normal(45, 5,  30)) +   # HR: average ₹45,000
        list(np.random.normal(70, 12, 30)) +   # Finance: average ₹70,000 (wide range)
        list(np.random.normal(80, 10, 30))     # IT: average ₹80,000
    )
})

# Show a preview of the data table
print("Data preview (first 5 rows):")
print(salary_data.head())
print(f"\nTotal rows: {len(salary_data)} (30 employees × 4 departments)")
```

```python
# ─── Draw the box plot ────────────────────────────────────────────

# data=   → the table we just created
# x=      → which column has the groups (departments)
# y=      → which column has the numbers (salaries)
sns.boxplot(data=salary_data, x='Department', y='Salary_K', palette='Set2')

plt.title('Salary Distribution by Department')
plt.xlabel('Department')
plt.ylabel('Monthly Salary (₹ Thousands)')

plt.show()

# ─── HOW TO READ THE CHART ────────────────────────────────────────
print("""
How to read this chart:
  → The BOX shows where the middle 50% of employees fall
  → The LINE inside the box = the median salary (middle person)
  → The WHISKERS (lines sticking out) = full typical range
  → Any DOTS outside the whiskers = outlier salaries (unusual)
  → TALL box = big salary gap within that department
  → SHORT box = most employees earn similar amounts
""")
```

```python
# ─── Step 1: Create some business data ───────────────────────────

import numpy as np
np.random.seed(0)

# Simulated monthly data for 12 months
business_data = pd.DataFrame({
    'Ad_Spend':   np.random.randint(10000, 50000, 12),    # Money spent on ads
    'Footfall':   np.random.randint(500,   2000,  12),    # Customers who visited store
    'Discount':   np.random.randint(5,     25,    12),    # Discount percentage given
    'Revenue':    np.random.randint(100000,500000,12),    # Total revenue earned
    'Returns':    np.random.randint(10,    100,   12)     # Items returned by customers
})

print("Sample business data (first 4 months):")
print(business_data.head(4))
```

```python
# ─── Step 2: Calculate correlations ──────────────────────────────

# .corr() calculates how each column relates to every other column
# The result is a table of numbers between -1 and +1
correlation_table = business_data.corr()

print("Correlation table (values between -1 and +1):")
print(correlation_table.round(2))
```

```python
# ─── Step 3: Visualise the correlation table as a heatmap ────────

plt.figure(figsize=(7, 5))    # Set the size of the chart (width=7, height=5 inches)

sns.heatmap(
    correlation_table,
    annot=True,       # annot=True → show the number inside each coloured box
    fmt='.2f',        # fmt='.2f' → show 2 decimal places (e.g. 0.73)
    cmap='coolwarm',  # cmap = colour map: red=positive, blue=negative, white=0
    vmin=-1,          # minimum value on colour scale
    vmax=1,           # maximum value on colour scale
    linewidths=0.5    # thin lines between each cell in the grid
)

# ✏️ Try changing cmap to: 'Blues', 'RdYlGn', 'viridis', 'YlOrRd'

plt.title('How Business Metrics Relate to Each Other')
plt.tight_layout()
plt.show()

print("""
How to read this heatmap:
  → Dark RED box = strong positive relationship (+1)
  → Dark BLUE box = strong negative relationship (−1)
  → WHITE / light box = little or no relationship (near 0)
  → The diagonal is always 1.0 (every variable perfectly matches itself)
""")
```

```python
# ─── Saving a chart as an image file ─────────────────────────────

months = ['Jan', 'Feb', 'Mar', 'Apr']
sales  = [50000, 70000, 65000, 90000]

sns.lineplot(x=months, y=sales)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Revenue (₹)')

# Save FIRST — before plt.show()
# dpi=300 means high resolution (good for printing)
# bbox_inches='tight' prevents labels from being cut off
plt.savefig('monthly_sales_chart.png', dpi=300, bbox_inches='tight')

# Show AFTER saving
plt.show()

print("Chart saved as: monthly_sales_chart.png")
print("You can find this file in the same folder as your notebook.")
```

Part 2 Plotly


```python
# Install Plotly if needed (remove the # and run)
# !pip install plotly

# Import Plotly
import plotly.express as px    # px = plotly.express — the easy-to-use version

print("Plotly is ready!")

# ─── NOTE ──────────────────────────────────────────────────────────
# plotly.express (px) is the beginner-friendly version of Plotly
# It lets you make beautiful charts with very little code
```

```python
# ─── YOUR FIRST PLOTLY CHART — Simple line chart ─────────────────

# Step 1: Create a data table (DataFrame)
# Plotly works best when data is in a pandas table
sales_data = pd.DataFrame({
    'Month': ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun'],
    'Sales': [50000, 70000, 65000, 90000, 85000, 110000]
})

# Step 2: Create the chart
# Notice: we store the chart in a variable called 'fig'
fig = px.line(
    sales_data,          # the data table
    x='Month',           # column name for horizontal axis
    y='Sales',           # column name for vertical axis
    title='Monthly Sales — Interactive Chart'
)

# Step 3: Show it
# Notice: fig.show() — NOT plt.show() — this is Plotly's way
fig.show()

print("\n👆 Try hovering over the line above!")
```
