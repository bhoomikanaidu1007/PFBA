```python
import pandas as pd
df = pd.read_csv('cleaned_sales.csv')
```

```python
category_summary = (
df.groupby('Product_Category')['Revenue']
.agg(['sum', 'mean', 'count'])
.rename(columns={
'sum' : 'Total Revenue (₹)',
'mean' : 'Avg Order Value (₹)',
'count' : 'No. of Orders'
})
.sort_values('Total Revenue (₹)', ascending=False)
.round(2)
)
print(category_summary)
```

```python
category_summary['Revenue Share (%)'] = (
category_summary['Total Revenue (₹)']
/ category_summary['Total Revenue (₹)'].sum()
* 100
).round(2)
```
